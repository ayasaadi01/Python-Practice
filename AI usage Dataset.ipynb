```python
#since gender has different cases, lowercase all of them
df["Gender"]= df.Gender.str.lower()
```

```python
#take of yrs to have just float numbers
df["Age"] = df["Age"].str.replace("yrs", "", case=False, regex=False).str.strip()
df["Age"] = pnd.to_numeric(df["Age"])
```

```python
# to replace hrs and turn hours into floats
df["AI_Usage_Hours_Per_Day"] = df["AI_Usage_Hours_Per_Day"].str.replace("hrs", "", case=False, regex=False).str.strip()
df["AI_Usage_Hours_Per_Day"] = pnd.to_numeric(df["AI_Usage_Hours_Per_Day"])
```

```python
# to change hrs by taking out the word "hrs" and turning time saved into float numbers
df["Time_Saved_Hours_Per_Week"] = df["Time_Saved_Hours_Per_Week"].str.replace("hrs", "", case=False, regex=False).str.strip()
df["Time_Saved_Hours_Per_Week"] = pnd.to_numeric(df["Time_Saved_Hours_Per_Week"])
```

```python
# to change hrs by taking out the word "hrs" and turning time saved into float numbers
df["Work_or_Study_Hours_Per_Day"] = df["Work_or_Study_Hours_Per_Day"].str.replace("hrs", "", case=False, regex=False).str.strip()
df["Work_or_Study_Hours_Per_Day"] = pnd.to_numeric(df["Work_or_Study_Hours_Per_Day"]) 
```

```python
# see if there are any duplicates in the rows, still some data cleaning to do
print(df.duplicated())
```

```python
# check gender values
df["Gender"].value_counts()
```

```python
# replace m with male, f for female and include non-binary as other
df["Gender"] = df["Gender"].str.lower().replace({"m": "male", "f": "female","non-binary": "other"})

df["Gender"].value_counts()
```

```python
# check all education values
df["Education_Level"].value_counts()
```

```python
# education map to have uniform data
education_mapping = {
    "Bachelor's": "Bachelor's",
    "Undergraduate": "Bachelor's",
    "undergrad": "Bachelor's",
    "BA/BSc": "Bachelor's",
    "Bachelors": "Bachelor's",
    "bachelor's degree": "Bachelor's",

    "Master's": "Master's",
    "Masters": "Master's",
    "MA/MSc": "Master's",
    "master's degree": "Master's",

    "PhD": "PhD",
    "Ph.D.": "PhD",

    "High School": "High School",
    "high school": "High School",
    "HS": "High School",

    "Diploma": "Diploma"
}
```

```python
# group by education level
df["Education_Level"] = df["Education_Level"].replace(education_mapping)
df["Education_Level"].value_counts()
```

```python
# lowercase country names to make them comparable
df["Country"]= df["Country"].str.lower()
```

```python
# check different countries' spelling
df["Country"].value_counts()
```

```python
# replace UK and USA
df["Country"] = df["Country"].str.lower().replace({"u.k.":"united kingdom", "united states":"united states of america"})
```

```python
# check if everything looks good
df["Country"].value_counts()
```

```python
# check different AI tools
df["AI_Tool"].value_counts()
```

```python
# group by AI tool
AI_tools_mapping = {
"google gemini" : "gemini",
"microsoft copilot" : "copilot",
"perplexity ai" : "perplexity",
"chat gpt" : "chatgpt"
}
```

```python
# lowercase and count
df["AI_Tool"] = df["AI_Tool"].str.lower()
df["AI_Tool"] = df["AI_Tool"].replace(AI_tools_mapping)
df["AI_Tool"].value_counts()
```

```python
# create a copy of profession to be able to compare the two columns, without losing important data
df["Profession_Original"] = df["Profession"].copy()
```

```python
# check professions and see potential grouping possibilities
df["Profession_Original"].value_counts()
```

```python
# create a lambda to group professions in macro areas
df["Profession"] = df["Profession_Original"].apply(
    lambda x: 
        "Student" if "student" in str(x).lower()
        else "Software Developer" if any(word in str(x).lower() for word in ["software", "developer","system","développeur","sistemas"])
        else "Customer Support" if any(word in str(x).lower() for word in ["customer", "support","cliente","kunden"])
        else "Analysts" if any(word in str(x).lower() for word in ["analista", "analyst","pesquisador"])
        else "Content Creation" if any(word in str(x).lower() for word in ["content", "producer","writer","contenu","marketing"])
        else "Academia" if any(word in str(x).lower() for word in ["research","academia"])
        else "Administration" if any(word in str(x).lower() for word in ["administrative","administrativo","manager"])
        else "Data Science" if any(word in str(x).lower() for word in ["data","machine learning","dados"])
        else "Other"
)
```

```python
# check that the professions have been grouped
df["Profession"].value_counts()
```

```python
# check original column against manipulated column, to see "Other" jobs that can be grouped in another way
df.loc[
    df["Profession"] == "Other",
    "Profession_Original"
].value_counts()
```
